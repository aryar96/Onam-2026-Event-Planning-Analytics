                                          Onam 2026 Event Registration + Lucky Draw System

```python
import random


events = {
    1: "Pookalam Competition",
    2: "Onam Games",
    3: "Thiruvathira",
    4: "Music Program",
    5: "Onam Sadya",
    6: "Traditional Dress Contest",
    7: "Cultural Program",
    8: "Thiruvonam Celebration"
}


participants = []


def register_participant():

    print("\n--- Register Participant ---")

    participant_id = int(input("Enter Participant ID: "))

    for participant in participants:
        if participant["id"] == participant_id:
            print("❌ Participant ID already exists.")
            return

    name = input("Enter Name: ")
    department = input("Enter Department: ")
    age = int(input("Enter Age: "))
    gender = input("Enter Gender: ")

    print("\nAvailable Events:")

    for key, value in events.items():
        print(key, "-", value)

    choice = int(input("Select Event: "))

    if choice not in events:
        print("❌ Invalid event selection.")
        return

    event_name = events[choice]

    participant = {
        "id": participant_id,
        "name": name,
        "department": department,
        "age": age,
        "gender": gender,
        "event": event_name
    }

    participants.append(participant)

    print("\n✅ Registration successful!")


def view_participants():

    print("\n--- Registered Participants ---")

    if not participants:
        print("No participants registered yet.")
        return

    for participant in participants:
        print(
            participant["id"],
            "|",
            participant["name"],
            "|",
            participant["department"],
            "|",
            participant["event"]
        )


def search_participant():

    print("\n--- Search Participant ---")

    participant_id = int(input("Enter Participant ID: "))

    for participant in participants:

        if participant["id"] == participant_id:

            print("\nParticipant Found ✅")
            print("ID:", participant["id"])
            print("Name:", participant["name"])
            print("Department:", participant["department"])
            print("Age:", participant["age"])
            print("Gender:", participant["gender"])
            print("Event:", participant["event"])

            return

    print("❌ Participant not found.")


def event_statistics():

    print("\n--- Event-wise Registration ---")

    event_count = {}

    for participant in participants:

        event = participant["event"]

        if event in event_count:
            event_count[event] += 1
        else:
            event_count[event] = 1

    if not event_count:
        print("No participants registered yet.")
        return

    for event, count in event_count.items():
        print(event, ":", count)


def lucky_draw():

    print("\n🎉 --- ONAM LUCKY DRAW --- 🎉")

    if not participants:
        print("No participants available.")
        return

    winner = random.choice(participants)

    print("\n🎊 Congratulations! 🎊")
    print("--------------------------")
    print("Winner :", winner["name"])
    print("ID     :", winner["id"])
    print("Event  :", winner["event"])
    print("--------------------------")
    print("🌼 Happy Onam 2026! 🌼")


while True:

    print("\n")
    print("🌼 ===============================")
    print("      ONAM 2026 CELEBRATION")
    print("      REGISTRATION SYSTEM")
    print("=============================== 🌼")

    print("1. Register Participant")
    print("2. View Participants")
    print("3. Search Participant")
    print("4. Event-wise Statistics")
    print("5. Lucky Draw")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        register_participant()

    elif choice == "2":
        view_participants()

    elif choice == "3":
        search_participant()

    elif choice == "4":
        event_statistics()

    elif choice == "5":
        lucky_draw()

    elif choice == "6":
        print("🌼 Thank you! Happy Onam 2026! 🌼")
        break

    else:
        print("❌ Invalid choice.")
```
